## Read a local CSV file

The sample file holds invented measurements for two plants.
Its path stays `data/measurements.csv` in the browser and in the download.


```python
import pandas as pd

measurements = pd.read_csv("data/measurements.csv")
measurements
```

## Draw the measurements

Seaborn draws one line for each plant.
The first code cell in the browser copy installs the packages found in the imports.


```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(7, 3.5))
sns.lineplot(data=measurements, x="week", y="height_cm", hue="plant", marker="o", ax=ax)
ax.set(xlabel="Week", ylabel="Height (cm)", title="Example plant measurements")
ax.set_xticks([1, 2, 3, 4])
fig.tight_layout()
plt.show()
```
